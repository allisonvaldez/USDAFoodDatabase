# USDA Food Database
This is an analysis that delves into the database provided by the US Department of Agriculture. It provides nutritional information for a plethora of food options in JSON format. If there are any questions or concerns the author may be contacted to discuss analysis and methodology.

```python
# Start with imports
import json
import pandas as pd
```

Create a database from the json file: 

```python
db = json.load(open('database.json'))
```

```python
# Preview the basics of the data
len(db)
```

Find the keys at [0]: 

```python
db[0].keys()
```

Now find the nutrients key value pairs:

```python
db[0]['nutrients'][0]
```

Create a DataFrame called nutrients that holds a dictionary of nutrients held at db[0]: 

```python
nutrients = pd.DataFrame(db[0]['nutrients'])
```

Gather nutritional data from our records into a single DataFrame where the rows link to the original nutitional information from the id column: 

```python
nutrients = []
for rec in db:
    fnuts = pd.DataFrame(rec['nutrients'])
    fnuts['id'] = rec['id']
    nutrients.append(fnuts)
nutrients = pd.concat(nutrients, ignore_index=True)
```

```python
# Preview a bit of the data
nutrients[:7]
```

Convert the list of dicts to a dataframe, making sure to specify the list of fields to extract:

```python
info_keys = ['description', 'group', 'id', 'manufacturer']
```

```python
# Create a new DataFrame 
info = pd.DataFrame(db, columns=info_keys)
```
